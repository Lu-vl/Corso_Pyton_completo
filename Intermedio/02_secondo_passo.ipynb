```python
import pandas as pd
# Creazione del dataframe(studenti) = dataset(tabella) 
studenti = {
    "nome": ["Mario", "Rudy", "Marco", "Roberta", "Anna", "Silvia"],
    "cognome": ["Ricci", "Botosso", "Secci", "Rossi", "Bianchi", "Verdi"],
    "eta": [30, 54, 54, 44, 23, 55],
    "citta": ["Milano", "Biella", "San Donà di Piave", "Napoli", "Roma", "Bari"],
    "voto": [8, 9, 7, 9, 4, 6],
    "stipendio": [2000, 3000, 4000, 2500, 1000, 3500]
}

# chiama pandas e crea un dataframe a partire dal dizionario studenti
df = pd.DataFrame(studenti)
df # stampa il dataframe(tipo tabella) con i dati degli studenti
print(df) # stampa il dataframe(tipo tabella in formato csv)
```

Filtri in pandas

```python
df[df["stipendio"] >= 2500] #select * from dipendenti where stipendio >= 2500
```

```python
#select stipendio
df[df["nome"] == "Giovanni"]["stipendio"]
```

```python

df[df['stipendio'] == 2500][['nome', 'cognome']]
```

```python
df[df['nome'].str.endswith('a')]
#nomi che finiscono con a
```

```python
#colonna citta contiene questa lettera:
df[df['citta'].str.contains('n')]
```
